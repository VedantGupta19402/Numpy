groupby:to summarise data

Region wise total coursepricem

```python
df.groupby(['Region'])['CoursePrice'].sum()
```

courseName wise toatal courseprice

```python
df.groupby(['CourseName'])['CoursePrice'].sum()
```

coursenamewise avg ratinf

```python
df.groupby(['CourseName'])['Rating'].mean()
```

Coursename wise min courseprice

```python
df.groupby(['CourseName'])['CoursePrice'].min()
```

```python
df.groupby(['CourseName'])['CoursePrice'].count()
```

Region and gender wise total courseprice

```python
df.groupby(['Region','Gender'])['CoursePrice'].sum()
```

payment methodwiser avg and total rating

```python
df.groupby(['PaymentMethod'])['Rating'].agg(['sum','mean']) ##2 opreations sath me lg gye ek sath
```

Payment method wise avg rating and totoal courseprice

```python
df.groupby(['PaymentMethod']).agg(
    avg_Rating=('Rating','mean'),
    total_sales=('CoursePrice','sum')
).reset_index()
```

pivot table:to summarise the data by rows and coloumns

```python
df.head()
```

region and gender wise avg rating

```python
df.pivot_table(index='Region',columns='Gender',values='Rating',aggfunc='mean')
```

paymentmethod and region wise total courseprice

```python
df.pivot_table(index='Region',columns='PaymentMethod',values='CoursePrice',aggfunc='mean')
```

regionwise avg and totalcourseprice

```python
df.pivot_table(index='Region',values='CoursePrice',aggfunc=['mean','sum'])
```

Region wise total courseprice and avg rating

```python
df.pivot_table(index='Region',values=['CoursePrice','Rating'],
               aggfunc={'CoursePrice':'sum','Rating':'mean'}
               ).rename(columns={'CoursePrice':'total','Rating':'Average'})
```
