```python
num=['Kilometers_Driven','Mileage_value','Engine_value','Power_value'] ##median se bhara isko outliers he
filled=['company_name','Price','Model_name','Fuel_type']
```

```python
df
```

```python
cat=[]
for i in df.columns:
    if i not in num and i not in filled:
        cat.append(i)
cat
```

Filling the null values

```python
cat
```

```python
for i in cat:
  df[i]=  df[i].fillna(df[i].mode()[0])
```

```python
for i in cat:
    print(i,":",df[i].isnull().sum())
```
